```python
import numpy as np
import pandas as pd

# load dataset
df = pd.read_csv("airbnb_listings_zurich.csv")
display(df.head())
print(df.columns)
```

```python
# drop colums with more than 30% missing values
missing_values = df.isnull().sum()
missing_percent = (missing_values / len(df)) * 100

cols_to_drop = missing_percent[missing_percent > 30]
print(cols_to_drop)
df = df.drop(columns=cols_to_drop.index)

# drop unessecary colums
url_cols = [col for col in df.columns if "url" in col.lower()]
df = df.drop(columns = url_cols + ["scrape_id", "source", "host_name", "host_verifications", "bathrooms_text"])

```
