### Housing Stock

```python
# load housing stock data set
stock_df = pd.read_csv("data/bau522od5221_wohnungsbestand_zurich.csv")

# only use entries from 2025
stock_df = stock_df[stock_df["StichtagDatJahr"] == 2025]
print(f'Shape of Housing Stock Dataset: {stock_df.shape}')
display(stock_df.head())
```

Since we are interested in the number of apartments in each neighbourhood, it would be convenient to have the data frame structured with each neighbourhood occupying one row and the number of apartments of different sizes to be the variables, including one stating the total number of apartments in the corresponding neighbourhood.

```python
pivot = stock_df.pivot_table(
    index='QuarLang',
    columns='AnzZimmerLevel2Lang_noDM',
    values='AnzWhgStat',
    aggfunc='sum'
)

# Total berechnen
pivot['Total'] = pivot.sum(axis=1)

display(pivot.head(22))
```
